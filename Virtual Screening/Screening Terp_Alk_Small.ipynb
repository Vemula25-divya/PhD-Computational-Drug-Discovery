```python
from chembl_webresource_client.new_client import new_client
import pandas as pd
import time

# Define all top candidates
candidates = {
    # Terpenoids
    'AGOSTEROL D2'                    : 'Terpenoid',
    'EUPHORBIASTEROID'                : 'Terpenoid',
    'CAMPHORATIN A'                   : 'Terpenoid',
    'PACLITAXEL CERIBATE'             : 'Terpenoid',
    'TAXOL C'                         : 'Terpenoid',
    'PHRYGIASTEROL'                   : 'Terpenoid',
    'AGOSTEROL C2'                    : 'Terpenoid',
    'AGOSTEROL G'                     : 'Terpenoid',
    'BETULINIC ACID SULFATE'          : 'Terpenoid',
    'PACLITAXEL DOCOSAHEXAENOIC ACID' : 'Terpenoid',
    # Alkaloids
    'TUBASTRINDOLES B'                : 'Alkaloid',
    'ERGOTAMINE'                      : 'Alkaloid',
    'TUBASTRINDOLES A'                : 'Alkaloid',
    'METERGOTAMINE'                   : 'Alkaloid',
    'DIHYDROERGOTAMINE MESYLATE'      : 'Alkaloid',
    'ACTINOMYCIN G6'                  : 'Alkaloid',
    'ACTINOMYCIN G4'                  : 'Alkaloid',
    'ACTINOMYCIN G3'                  : 'Alkaloid',
    'ACTINOMYCIN Z1'                  : 'Alkaloid',
    'ACTINOMYCIN G2'                  : 'Alkaloid',
    # Fragments
    'CHEMBL2012186'                   : 'Fragment',
    'CHEMBL422438'                    : 'Fragment',
    'SULBACTAM BENZATHINE'            : 'Fragment',
    'CHEMBL1999780'                   : 'Fragment',
    'RITLECITINIB TOSYLATE'           : 'Fragment',
    'CHEMBL1996491'                   : 'Fragment',
    'CHEMBL3502665'                   : 'Fragment',
    'CHEMBL310274'                    : 'Fragment',
    'CHEMBL1916025'                   : 'Fragment',
    'CODEINE PHOSPHATE'               : 'Fragment',
}

print(f"Total candidates: {len(candidates)}")
print("Fetching SMILES from ChEMBL...")

molecule = new_client.molecule
smiles_data = []

for name, library in candidates.items():
    try:
        # Search by name
        result = molecule.search(name)
        if result:
            for r in result:
                try:
                    smiles = r['molecule_structures']['canonical_smiles']
                    chembl_id = r['molecule_chembl_id']
                    mw = r['molecule_properties']['mw_freebase']
                    logp = r['molecule_properties']['alogp']
                    if smiles:
                        smiles_data.append({
                            'name'       : name,
                            'chembl_id'  : chembl_id,
                            'library'    : library,
                            'MW'         : mw,
                            'LogP'       : logp,
                            'smiles'     : smiles
                        })
                        break
                except:
                    continue
    except:
        print(f"  Could not fetch: {name}")
    time.sleep(0.1)  # avoid API overload

# Convert to DataFrame
docking_new = pd.DataFrame(smiles_data)
docking_new = docking_new.drop_duplicates(
    subset=['name']
).reset_index(drop=True)

# Add ML scores
terp_df  = pd.read_csv('screening_results/terpenoid_compounds_top20.csv')
alk_df   = pd.read_csv('screening_results/alkaloid_compounds_top20.csv')
frag_df  = pd.read_csv('screening_results/fragment_library_top20.csv')

all_scores = pd.concat([terp_df, alk_df, frag_df],
                        ignore_index=True)

docking_new = docking_new.merge(
    all_scores[['drug_name', 'active_probability',
                'predicted_pIC50']],
    left_on='name',
    right_on='drug_name',
    how='left'
).drop(columns=['drug_name'])

# Reorder columns
docking_new = docking_new[[
    'name', 'chembl_id', 'library',
    'MW', 'LogP',
    'active_probability', 'predicted_pIC50',
    'smiles'
]]

# Save
docking_new.to_csv(
    'glide_docking/new_docking_candidates.csv', index=False
)

print(f"\nDocking CSV saved ✅")
print(f"Location: glide_docking/new_docking_candidates.csv")
print(f"Total compounds: {len(docking_new)}")
print(f"\nFull table:")
print(docking_new.to_string(index=False))
```
